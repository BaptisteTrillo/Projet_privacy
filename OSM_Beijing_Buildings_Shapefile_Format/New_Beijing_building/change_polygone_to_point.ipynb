```python
import geopandas as gpd

# Lire le fichier Shapefile
gdf = gpd.read_file("../OSM_Beijing_Buildings_Shapefile_Format/gis_osm_buildings_a_free_1.shp")

# Afficher les 5 premières lignes
print(gdf.head())

# Vérifier le système de coordonnées (CRS)
print(gdf.crs)

# Vérifier le nombre de lignes
print(f"Nombre d'entités : {len(gdf)}")
```

```python
# Vérifier le système de coordonnées (CRS)
print("CRS original :", gdf.crs)

# Reprojecter les données en UTM Zone 50N (EPSG:32650)
gdf = gdf.to_crs('EPSG:32650')

# Calculer les centroïdes et remplacer la géométrie
gdf['geometry'] = gdf.geometry.centroid
gdf['name'] = None
# Afficher les 5 premières lignes des données avec les centroïdes
print("Données avec centroïdes :")
print(gdf.head())

# Vérifier le nombre de lignes
print(f"Nombre d'entités : {len(gdf)}")

# Vérifier le CRS après reprojection
print("CRS après reprojection :", gdf.crs)
```

```python
# Enregistrer le GeoDataFrame dans un fichier Shapefile
output_path = "../../my-home-is-my-secret/my-home-is-my-secret-master/input/beijing_buildings_centroids.shp"
gdf.to_file(output_path)



print(f"Fichier Shapefile enregistré sous : {output_path}")
```

```python
# Relire le fichier Shapefile pour vérification
gdf_check = gpd.read_file(output_path)

# Afficher les 5 premières lignes
print(gdf_check.head())

# Vérifier le CRS
print("CRS du fichier Shapefile :", gdf_check.crs)
```

```python
import pandas as pd
trajectory = gpd.read_file("../../my-home-is-my-secret/my-home-is-my-secret-master/input/synthetic_trajectories_hel.shp")

print(trajectory.head())

# Vérifier le nombre de lignes
print(f"Nombre d'entités : {len(trajectory)}")

# Créer un nouveau DataFrame avec les colonnes 'id' et 'geometry_wkt'
new_trajectory = pd.DataFrame({
    'id': trajectory['id'],  # Conserver la colonne 'id'
    'geometry_wkt': trajectory['geometry'].apply(lambda geom: geom.wkt)  # Convertir les géométries en WKT
})

# Afficher le nouveau DataFrame
print(new_trajectory.head())

# 3. Exporter en CSV
output_trajectory_csv = "trajectory_finland.csv"
new_trajectory.to_csv(output_trajectory_csv, index=False, encoding='utf-8')
```

```python
from shapely.ops import nearest_points

# Fonction pour calculer la distance de Hausdorff entre deux géométries
def hausdorff_distance(geom1, geom2):
    return geom1.hausdorff_distance(geom2)

# Trouver la correspondance pour chaque trajectoire tronquée
matches = []
for _, truncated_row in truncated.iterrows():
    truncated_geom = truncated_row['geometry']
    best_match_id = None
    best_distance = float('inf')
    # Comparer avec toutes les trajectoires originales
    for _, original_row in trajectory.iterrows():
        original_geom = original_row['geometry']
        distance = hausdorff_distance(truncated_geom, original_geom)
        
        # Garder la meilleure correspondance
        if distance < best_distance:
            best_distance = distance
            best_match_id = original_row['id']
    
    # Ajouter la correspondance à la liste
    matches.append((truncated_row.name, best_match_id, best_distance))  # truncated_row.name est l'index de la trajectoire tronquée

# Créer un DataFrame pour les correspondances
matches_df = pd.DataFrame(matches, columns=['truncated_index', 'original_id', 'hausdorff_distance'])

# Afficher les correspondances
print(matches_df.head())
```
